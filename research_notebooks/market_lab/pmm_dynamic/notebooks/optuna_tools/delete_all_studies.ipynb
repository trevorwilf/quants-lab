```python
# ── Optuna Study Manager ──
# Run this cell to list, delete one, or delete all studies.
# Works with both SQLite (optuna_studies.db) and Postgres (OPTUNA_STORAGE env).

import optuna
import os

from pmm_lab.optuna.storage import get_storage_url

storage_url = get_storage_url()
print(f"Storage: {storage_url}\n")

# ── List all studies ──
studies = optuna.study.get_all_study_summaries(storage=storage_url)
if not studies:
    print("No studies found.")
else:
    print(f"Found {len(studies)} study/studies:\n")
    for i, s in enumerate(studies):
        n_trials = s.n_trials
        best = f"{s.best_trial.value:.4f}" if s.best_trial else "N/A"
        print(f"  [{i}] {s.study_name}  —  {n_trials} trials, best={best}")

```
